```python
# USSD App Simulation - GTBank
# Uses variables, loops, and functions

balance = 25000
pin = '1234'

def check_balance():
    return f"Your balance is ₦{balance}"

def transfer_money(amount, recipient):
    global balance
    if amount > balance:
        return "Insufficient funds"
    balance -= amount
    return f"₦{amount} sent to {recipient}. New balance: ₦{balance}"

def buy_airtime(amount, phone_number):
    global balance
    if amount > balance:
        return "Insufficient funds"
    balance -= amount
    return f"₦{amount} airtime sent to {phone_number}"

def ussd_menu():
    entered_pin = input("Enter your 4-digit PIN: ")
    if entered_pin != pin:
        print("Incorrect PIN")
        return

    while True:
        print("\n--- GTBank USSD Menu ---")
        print("1. Check Balance")
        print("2. Transfer Money")
        print("3. Buy Airtime")
        print("4. Exit")

        choice = input("Select an option: ")

        if choice == "1":
            print(check_balance())
        elif choice == "2":
            amt = float(input("Enter amount: "))
            recipient = input("Enter recipient name: ")
            print(transfer_money(amt, recipient))
        elif choice == "3":
            amt = float(input("Enter amount: "))
            phone = input("Enter phone number: ")
            print(buy_airtime(amt, phone))
        elif choice == "4":
            print("Thank you for using GTBank USSD.")
            break
        else:
            print("Invalid option, try again.")

ussd_menu()
```
